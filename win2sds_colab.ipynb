# WIN to SDS (Google Colab)

Convert WIN-format seismic data to miniseed in SeisComP Data Structure (SDS). Run the cell below; you will be prompted to **upload** a zip of your WIN directory and a lookup CSV. After conversion, **sds_output.zip** will download to your machine.

---

## Suggested file sizes

| What | Limit |
|------|--------|
| Upload (zip + CSV) | ≤ 1–2 GB total |
| Download (result zip) | ≤ 2–3 GB |

Larger uploads often time out. For bigger data, use Google Drive (mount Drive, then use `win2sds` with paths under `/content/drive/MyDrive/...`).

---

## Inputs

- **WIN zip:** A zip of the folder that contains your WIN subfolders (e.g. `yymmddHH/yymmddHH.01`). On Windows your source might be `C:\Users\You\win_data\` (zip that folder).
- **Lookup CSV:** Columns `win`, `mseed`; optional `lat`, `lon`, `elev`. Lines starting with `#` are skipped. On Windows you might have `C:\Users\You\lookup.csv`.

Example CSV:

```
win,mseed,lat,lon,elev
ac60,VG.SKY.00.EHZ,1.245,127.939,450
```

**Example paths (Windows):** On your machine you might have `C:\Users\You\win_data\` (WIN folder to zip), `C:\Users\You\lookup.csv`, and `C:\Users\You\SDS` (where you’ll unzip the result). In Colab the code below uses `/content/...` paths; the upload step brings your files into Colab.
